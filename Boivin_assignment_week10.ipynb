1. Count Vowels

Write a function `count_vowels(s)` that takes a string and returns the number of vowels in it (`a`, `e`, `i`, `o`, `u` — both upper and lower case).

```python
>>> count_vowels("hello")
2
>>> count_vowels("HEllO World")
3

````python
def count_vowels(text):

  ''' count_vowels(string) --> integer

  count_vowels will take any string and return the number of vowels in that string

  >>> count_vowels("hello") == 2
  >>> count_vowels("HELLO") == 2
  >>> count_vowels("xyz") == 0
  '''

  count = 0
  vowels = 'aAeEiIoOuU'

  for character in text:
    if character in vowels:
        count += 1
  return count
````

````python
count_vowels("hEllo")
````

````python
import doctest
doctest.run_docstring_examples(count_vowels, globals(), verbose=True)
````

````python
assert count_vowels("hello") == 2
assert count_vowels("HELLO") == 2
assert count_vowels("xyz") == 0
````

3 Filter Short Words

Create a function filter_short(words, max_len) that returns a list of only those words whose length is less than or equal to max_len.
```python
>>> filter_short(['hi', 'hello', 'no', 'yes'], 2)
['hi', 'no']

````python
def filter_short(words, max_len):

  ''' filter_short (['string'], integer) --> ['string']

  filter_shorts will take any string and will return a list of string that are less than or equal to the max number of characters or len.

  >>> filter_short(['computers', 'science', 'sunny', 'JIMmy'], 5)
  >>> filter_short (['APPLEs', 'paIr', 'GraPes', 'tomato', 'AcorN'], 4)
  '''

  word_list = []

  for word in words:
      if len(word) <= max_len:
          word_list.append(word)

  return word_list
````

````python
doctest.run_docstring_examples(filter_short, globals(), verbose = True)
````

````python
print(filter_short(['hi', 'hello', 'no', 'yes'], 2))
````
